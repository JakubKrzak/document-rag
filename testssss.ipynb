```python
from docling.datamodel.document import DoclingDocument
from docling.chunking import HybridChunker
from app.services.file_parsed_content import parsed_file_content
import asyncio
```

```python
path = "/home/kuba/Desktop/rag/disc/programowanie_d57ad736.pdf"
document = await parsed_file_content(path)
```

```python

async def chunk_document(parsed_document: DoclingDocument) -> list[DoclingDocument]:
    chunker = HybridChunker() 
    chunks = await asyncio.to_thread(
        lambda: list(chunker.chunk(parsed_document))
    )
    return chunks


chunks = await chunk_document(document)
```

```python


from docling.datamodel.document import DoclingDocument
from docling_core.transforms.chunker import BaseChunker
from docling.chunking import HybridChunker
from app.services.file_parsed_content import parsed_file_content
import asyncio
import json
from pathlib import Path
```

```python
def save_chunks_on_disc(chunks: list[BaseChunker]):
    if not chunks:
        return []
    
    name = Path(chunks[0].meta.origin.filename)
    path = f"chunks_disc/{name.stem}.jsonl"

    with open(path, "w", encoding="utf8") as f:
        for chunk in chunks:
            f.write(json.dumps(chunk.model_dump(), ensure_ascii=False) + "\n")
```

```python

save_chunks_on_disc(chunks)
```
